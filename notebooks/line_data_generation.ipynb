```python
import matplotlib.pyplot as plt

# Sample data
x = ['jan', 'feb', 'mar', 'apr', 'may', 'jun', 'jul', 'aug', 'sep', 'oct', 'nov', 'dec']  # X values ranging from 1 to 10
y = [2, 3, 3, 8, 6, 12, 8, 10, 10, 11, 9, 13]  # Y values, each one unit greater than its corresponding X value

# Creating the plot
plt.plot(x, y)

# Adding title
plt.title('Sample Line Plot')

# Adding X and Y labels
plt.xlabel('X Axis Label')
plt.ylabel('Y Axis Label')

# Showing the plot
plt.show()
```

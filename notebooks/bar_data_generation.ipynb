```python
import random
import pandas as pd
```

```python
n_times = random.randint(3, 10)
n_times
```

```python
base_number = random.randint(0, 200)
base_number
```

```python
lower_bound = base_number - 20
upper_bound = base_number + 20
```

```python
generated_numbers = []
for _ in range(n_times):
    number = random.randint(lower_bound, upper_bound)
    generated_numbers.append(number)
    
generated_numbers
```

```python
df = pd.DataFrame({f"Data_{base_number}": generated_numbers})
df
```

```python
full_df = pd.DataFrame()

for i in range(0, 200):
    n_times = random.randint(3, 10)
    
    base_number = random.randint(0, 200)
    
    lower_bound = max(5, base_number - 20)
    upper_bound = base_number + 20
    
    generated_numbers = [random.randint(lower_bound, upper_bound) for _ in range(n_times)]

    full_df[f"{i+1}"] = pd.Series(generated_numbers)
```

```python
full_df.shape
```

```python
full_df.to_csv('testData.csv', index=False)
```
